## Context 

When renting a car, our users have to complete a checkin flow at the beginning of the rental and a checkout flow at the end of the rental in order to:

* Assess the state of the car and notify other parties of pre-existing damages or damages that occurred during the rental.

The checkin and checkout of our rentals can be done with three distinct flows:
* **📱 Mobile** rental agreement on native apps: driver and owner meet and both sign the rental agreement on the owner’s smartphone

## Project 🚧

When using Getaround, drivers book cars for a specific time period, from an hour to a few days long. They are supposed to bring back the car on time, but it happens from time to time that drivers are late for the checkout.

Late returns at checkout can generate high friction for the next driver if the car was supposed to be rented again on the same day : Customer service often reports users unsatisfied because they had to wait for the car to come back from the previous rental or users that even had to cancel their rental because the car wasn’t returned on time.


## Goals 🎯

In order to mitigate those issues we’ve decided to implement a minimum delay between two rentals. A car won’t be displayed in the search results if the requested checkin or checkout times are too close from an already booked rental.

It solves the late checkout issue but also potentially hurts Getaround/owners revenues: we need to find the right trade off.

### Part des retards par seuil

````python
thresholds = [0, 15, 30, 60, 120]       #definition des seuls à etudier
late = delays.dropna(subset=["delay_at_checkout_in_minutes"])

share_late = pd.DataFrame({
    f"retard > {t} min (%)": late.groupby("checkin_type")["delay_at_checkout_in_minutes"].apply(lambda s: (s > t).mean() * 100)
    for t in thresholds
})
share_late.loc["total"] = [(late["delay_at_checkout_in_minutes"] > t).mean() * 100 for t in thresholds]
share_late.round(1)
````

Plus d'un conducteur sur deux rend la voiture en retard, mais seulement un quart dépasse une heure de retard et un sur six dépasse deux heures. La qualification de retour tardif dépend donc du seuil retenu.   
Nous observons également que les réservations de type connect conduisent à un pourcentage de retards plus faible que les réservations de type mobile.

Une option qui s'appliquerait aux cas "connect" seulement viserait precisement ceux qui sont le moins souvent en retard, d'autant plus que les reservations 'connect' sont moins nombreuses que les reservations 'mobile'. Une telle option se serait toutefois pas denuée de sens car la frequence n'est pas le seul critere à prendre en compte, mais la gravité du retard aussi. En "mobile" une personne est presente pour la remise des clefs et peut gerer la situation, coordoner le conducteur précedent avec le conducteur suivant, et transmettre une information. ### Construction des paires de locations consécutives

Dans cette partie, nous allons nous intéresser à l'effet d'un retard sur la location suivante. Nous allons donc devoir associer les locations par paires. Pour cela, nous allons filtrer la colonne previous_ended_rental_id du dataframe delays pour ne conserver que les lignes qui ne sont pas vides sur cette colone. Notons également que les éléments de la colonne previous_ended_rental_id doivent être convertis en entiers pour pouvoir être comparés à rental_id. En effet, la présence de NaN dans la colonne initiale apour effet de les convertir en decimaux.

#### Filtrage et conversion

````python
pairs = delays.dropna(subset=["previous_ended_rental_id"]).copy()           #on garde les 1841 locations qui ont une location precedente dans les 12h.
pairs["previous_ended_rental_id"] = pairs["previous_ended_rental_id"].astype(int)     #conversion des ID en entier.
````

#### Création de la table previous

Pour chaque élément du tableau `delays' nous disposons de l'information sur le retard d'un conducteur à la remise du véhicule, ainsi que La durée séparant sa réservation de la réservation précédente. Or, pour pouvoir faire une analyse de l'impact d'une réservation sur la suivante, nous avons besoin de mettre en regard la durée séparant la réservation de la réservation précédente avec le retard du conducteur precedent et non du conducteur en cours. Cette information peut être obtenue à partir de l'information existante, mais nécessite un travail de jointure entre la table 'delays' et une table que nous créerons sur mesure pour rendre cette jointure possible.

Nous allons donc creer un dataframe 'previous' qui reprend toutes les locations reduites à leur identifiant et à leur retard. 
Afin d'éviter la collision entre les noms de colonnes de ces deux tables, nous devons renommer les colonnes de previous. 

- rental_id sera renomée en previous_ended_rental_id. Ce nom de colonne existe deja et c'est precisement sur cette colonne que se fera la jointure. Le renommage ici n'est pas arbitraire.
- delay_at_checkout_in_minutes sera renommé en previous_delay

````python
previous = delays[["rental_id", "delay_at_checkout_in_minutes"]].rename(
    columns={"rental_id": "previous_ended_rental_id", "delay_at_checkout_in_minutes": "previous_delay"}
)
````

#### Jointure

La jointure a été préparée dans l'étape précédente. Nous n'avons plus qu'à l'appliquer ici. 

````python
pairs = pairs.merge(previous, on="previous_ended_rental_id")
print(pairs.shape)
print(pairs["previous_delay"].isna().sum())
pairs.head(10)
````

Le nouveau data frame pairs 1 841 lignes, dont 112 ont un previous_delay qui est inconnu.

# 4. Effet du délai minimum selon le seuil et le périmètre

Dans cette partie, nous souhaitons aider le Product Manager à décider de deux choses :

- la durée du délai minimum entre deux locations, que nous appellerons le seuil ;
- le type de réservation auquel il s'applique, connect seulement ou toutes les réservations (connect et mobile), que nous appellerons le périmètre.   

Pour cela, nous repassons en revue l'ensemble des paires de locations consécutives, soit 1 841 paires, et nous étudions tous les seuils de 0 à 720 minutes (soit 12 h) par incrément de 30 minutes, pour les deux périmètres (connect et all).   
Pour chaque seuil et chaque périmètre, nous comptons deux choses :

- les locations affectées : celles dont le créneau prévu commençait avant le seuil lié à la reservation précédente.
- les cas résolus : parmi ces locations refusées, celles où la voiture n'était effectivement pas revenue à l'heure du checkin.

Un cas problématique est donc considéré comme résolu si la règle aurait empêché la réservation.

Notons que la table contient 112 réservations pour lesquelles le retard du conducteur précédent est inconnu. Pour celles-ci, nous savons si la règle les aurait refusées, mais pas si elles ont subi un retard : elles comptent donc parmi les locations affectées, jamais parmi les cas résolus. C'est pourquoi elles sont conservées.

### Simulation sur les paires consécutives

````python
delay_grid = range(0, 721, 30)

def simulate(df, threshold):         # Definition d'une fonction de simulation 
    blocked = df["time_delta_with_previous_rental_in_minutes"] < threshold
    problematic = df["delay_for_next_driver"] > 0
    return {"threshold": threshold, "affected_rentals": blocked.sum(), "solved_cases": (blocked & problematic).sum()}

simulation = pd.concat([
    pd.DataFrame([simulate(pairs, t) for t in delay_grid]).assign(scope="all"),
    pd.DataFrame([simulate(pairs[pairs["checkin_type"] == "connect"], t) for t in delay_grid]).assign(scope="connect"),
])
simulation
````

Ce tableau nous permet de faire plusieurs observations. 

1) Les cas résolus plafonnent rapidement. Nous pouvons observer qu'au-delà de 170 180 cas résolus, Le nombre de résolutions ne croît plus fortement avec les locations affectées. Les cas resolus plafonnent à 216 au bout de 12 h, mais le nombre de 180 est atteint au bout de seulement 2 heures. On a donc plus de 80 % des cas résolus avec un seuil à 2 h. 666 locations auraient été affectées. 
2) Le nombre de cas de location affectés, lui, continue de croître sans plafonnement et de manière à peu près régulière. 
3) Un compromis semble se trouver entre 1 et 2 heures, où la majorité des cas seront résolus, tout en limitant le nombre de locations affectées
4) En connect seulement, le schéma semble à peu près identique, avec des cas résolus qui plafonnent rapidement dans les 2 heures et des locations affectées qui, elles, continuent de croître régulièrement. Le plafond est atteint à 69 au bout de 12 heures, mais 59 cas sont résolus au bout de 2 heures, soit 85 %. 295 locations auraient été affectées. 

### Représentation graphique

Nous allons faire une représentation graphique des des locations affectees et des cas resolus afin de verifier les observations visuellement. 

````python
fig = px.line(
    simulation[simulation["scope"] == "all"].melt(id_vars=["threshold", "scope"], var_name="indicator", value_name="count"),
    x="threshold",
    y="count",
    color="indicator",
    title="Locations affectées et cas résolus selon le seuil, toutes les réservations",
    labels={"threshold": "Seuil (minutes)", "count": "Nombre de locations", "indicator": ""},
)
fig.show()
````

````python
fig = px.line(
    simulation[simulation["scope"] == "connect"].melt(id_vars=["threshold", "scope"], var_name="indicator", value_name="count"),
    x="threshold",
    y="count",
    color="indicator",
    title="Locations affectées et cas résolus selon le seuil, réservations connect",
    labels={"threshold": "Seuil (minutes)", "count": "Nombre de locations", "indicator": ""},
)
fig.show()
````

Les graphiques confirment les observations faites avec le tableau. On observe effectivement un plafonnement des courbes en rouge représentant les cas résolus, mais une croissance relativement régulière de la courbe des courbes en bleu représentant les locations affectées. 

# 5. Part du revenu des propriétaires concernée

Les données ne précisent pas les prix de chaque location individuellement, et le revenu impacté ne peut pas être calculé directement. 
Sera donc analysé sous l'angle de la proportion des locations terminées qui aurait été refusées, rapportées à l'ensemble des locations terminées.

Ici, nous allons créer deux dataframes, l'un contenant l'ensemble des locations terminées (ended), en effet seules ces locations génèrent du revenu ; l'autre contenant les paires consécutives dont la location suivante s'est effectivement terminée (ended_pairs), c'est-à-dire les seules que la règle pourrait refuser en faisant perdre un revenu. Une fonction calcule ensuite, pour un seuil donné, la part des locations terminées que la règle aurait refusées, et la table l'applique à chaque seuil pour les deux périmètres.


````python
#locations terminées
ended = delays[delays["state"] == "ended"]

# paires consécutives dont la location suivante s'est terminée
ended_pairs = pairs[pairs["state"] == "ended"]

# part des locations terminées que la règle aurait refusées, pour un seuil donné
def revenue_share(df, threshold):
    return (df["time_delta_with_previous_rental_in_minutes"] < threshold).sum() / len(ended) * 100


# une ligne par seuil, une colonne par périmètre
pd.DataFrame({
    "threshold": delay_grid,
    "all (%)": [revenue_share(ended_pairs, t) for t in delay_grid],
    "connect (%)": [revenue_share(ended_pairs[ended_pairs["checkin_type"] == "connect"], t) for t in delay_grid],
}).round(2)
````

Dans ce tableau, les colonnes « all » et « connect » représentent le pourcentage de locations qui seraient affectées par la mise en place d'un seuil en rapport au nombre de locations terminées.   
Dans la zone de compromis identifiée plus haut, comprise entre 1 et 2 heures, le revenu en jeu est de 2 à 3 % sur l'ensemble des réservations et de moins de 1,5 % sur les réservations connectées seules. 

Le choix d'un seuil peut ainsi être réalisé en fonction du budget de perte assumées que le product manager est prêt à allouer à une plus grande fiabilité de la disponibilité des véhicules lors de l'arrivée du prochain conducteur.
